```python
import matplotlib.pyplot as plt
```

```python
scores = [68, 72, 75, 70, 78, 74, 76, 71]
prior_mean = 70
sample_mean = sum(scores) / len(scores)   # 73.0

# Posterior mean from Question 3
post_var = 1 / (1/16 + 8/25)
post_mean = post_var * (70/16 + 8*sample_mean/25)   # 72.51
```

```python
fig, ax = plt.subplots(figsize=(6.5, 3.4), dpi=200)
ax.bar(range(1, 9), scores, color='#4C78A8')

ax.axhline(prior_mean, ls='--', color='#E45756', lw=1.6, label='Prior mean = 70')
ax.axhline(sample_mean, ls='--', color='#333333', lw=1.6, label='Sample mean = 73')
ax.axhline(post_mean, ls='--', color='#54A24B', lw=1.6, label=f'Posterior mean = {post_mean:.2f}')

ax.set_ylim(60, 80)
ax.set_xlabel('Student')
ax.set_ylabel('Test score')
ax.set_title('Student test scores with prior, sample and posterior means')
ax.legend(fontsize=8, loc='upper left', ncol=3, frameon=False)
plt.tight_layout()

plt.savefig('chart.png')   # for your Word document
plt.show()                 # to view it on screen
```
